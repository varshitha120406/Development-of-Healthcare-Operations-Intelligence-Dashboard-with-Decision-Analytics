### Question 1: Reverse a String

Write a function that takes a string as input and returns the string reversed. For example, `reverse_string("hello")` should return `"olleh"`.

```python
def reverse_string(s):
    return s[::-1]

# Test cases
print(f"'hello' reversed is: {reverse_string('hello')}")
print(f"'python' reversed is: {reverse_string('python')}")
```

### Question 2: Find the First Non-Repeating Character

Given a string, find the first non-repeating character in it and return its index. If it doesn't exist, return -1. For example, for `s = "leetcode"`, return `0`. For `s = "loveleetcode"`, return `2`.

```python
from collections import Counter

def first_non_repeating_char(s):
    char_counts = Counter(s)
    for i, char in enumerate(s):
        if char_counts[char] == 1:
            return i
    return -1

# Test cases
print(f"First non-repeating char index for 'leetcode': {first_non_repeating_char('leetcode')}")
print(f"First non-repeating char index for 'loveleetcode': {first_non_repeating_char('loveleetcode')}")
print(f"First non-repeating char index for 'aabb': {first_non_repeating_char('aabb')}")
```

### Question 3: Check for Palindrome

Write a function that checks if a given string is a palindrome. A palindrome reads the same forwards and backwards. For example, `is_palindrome("madam")` should return `True`.

```python
def is_palindrome(s):
    cleaned_s = ''.join(filter(str.isalnum, s)).lower()
    return cleaned_s == cleaned_s[::-1]

# Test cases
print(f"'madam' is palindrome: {is_palindrome('madam')}")
print(f"'A man, a plan, a canal: Panama' is palindrome: {is_palindrome('A man, a plan, a canal: Panama')}")
print(f"'hello' is palindrome: {is_palindrome('hello')}")
```

### Question 4: FizzBuzz

Write a program that prints the numbers from 1 to 100. But for multiples of three print "Fizz" instead of the number and for the multiples of five print "Buzz". For numbers which are multiples of both three and five print "FizzBuzz".

```python
def fizzbuzz():
    for i in range(1, 101):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

# Call the function to print FizzBuzz sequence
fizzbuzz()
```

### Question 5: Two Sum

Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`. You may assume that each input would have exactly one solution, and you may not use the same element twice. For example, `nums = [2,7,11,15]`, `target = 9` should return `[0,1]`.

```python
def two_sum(nums, target):
    num_map = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]
        num_map[num] = i
    return [] # Should not reach here based on problem statement

# Test cases
print(f"Two sum for nums=[2,7,11,15], target=9: {two_sum([2,7,11,15], 9)}")
print(f"Two sum for nums=[3,2,4], target=6: {two_sum([3,2,4], 6)}")
```

### Question 6: Merge Two Sorted Lists

Merge two sorted linked lists and return it as a new sorted list. The new list should be made by splicing together the nodes of the first two lists.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def merge_two_lists(l1, l2):
    dummy = ListNode()
    current = dummy

    while l1 and l2:
        if l1.val < l2.val:
            current.next = l1
            l1 = l1.next
        else:
            current.next = l2
            l2 = l2.next
        current = current.next

    if l1:
        current.next = l1
    elif l2:
        current.next = l2

    return dummy.next

# Helper to convert list to ListNode
def list_to_linkedlist(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

# Helper to convert ListNode to list
def linkedlist_to_list(head):
    arr = []
    current = head
    while current:
        arr.append(current.val)
        current = current.next
    return arr

# Test cases
l1 = list_to_linkedlist([1, 2, 4])
l2 = list_to_linkedlist([1, 3, 4])
merged_list = merge_two_lists(l1, l2)
print(f"Merged list for [1,2,4] and [1,3,4]: {linkedlist_to_list(merged_list)}")

l3 = list_to_linkedlist([])
l4 = list_to_linkedlist([0])
merged_list_2 = merge_two_lists(l3, l4)
print(f"Merged list for [] and [0]: {linkedlist_to_list(merged_list_2)}")
```

### Question 7: Implement Factorial

Write a function to calculate the factorial of a non-negative integer. The factorial of a number `n` is the product of all positive integers less than or equal to `n`. For example, `factorial(5)` should return `120`.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")
    if n == 0:
        return 1
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# Test cases
print(f"Factorial of 5: {factorial(5)}")
print(f"Factorial of 0: {factorial(0)}")
print(f"Factorial of 7: {factorial(7)}")
```

### Question 8: Find the Missing Number

Given an array `nums` containing `n` distinct numbers in the range `[0, n]`, return the only number in the range that is missing from the array. For example, `nums = [3,0,1]` should return `2`.

```python
def find_missing_number(nums):
    n = len(nums)
    # Sum of numbers from 0 to n
    expected_sum = n * (n + 1) // 2
    # Sum of numbers in the given array
    actual_sum = sum(nums)
    return expected_sum - actual_sum

# Test cases
print(f"Missing number in [3,0,1]: {find_missing_number([3,0,1])}")
print(f"Missing number in [9,6,4,2,3,5,7,0,1]: {find_missing_number([9,6,4,2,3,5,7,0,1])}")
print(f"Missing number in [0]: {find_missing_number([0])}")
```

### Question 9: Validate Parentheses

Given a string `s` containing just the characters `'('`, `')'`, `'{'`, `'}'`, `'['` and `']'`, determine if the input string is valid. An input string is valid if: Open brackets must be closed by the same type of brackets. Open brackets must be closed in the correct order. Every close bracket has a corresponding open bracket of the same type.

```python
def is_valid_parentheses(s):
    stack = []
    mapping = {"(": ")", "{": "}", "[": "]"}
    open_brackets = set(mapping.keys())

    for char in s:
        if char in open_brackets:
            stack.append(char)
        elif stack and mapping[stack[-1]] == char:
            stack.pop()
        else:
            return False

    return not stack

# Test cases
print(f"'()' is valid: {is_valid_parentheses('()')}")
print(f"'()[]{}' is valid: {is_valid_parentheses('()[]{}')}")
print(f"'(]' is valid: {is_valid_parentheses('(]')}")
print(f"'([{}])' is valid: {is_valid_parentheses('([{}])')}")
print(f"'{[)}' is valid: {is_valid_parentheses('{[})')}")
```

### Question 10: Implement a Queue using Two Stacks

Implement a first in first out (FIFO) queue using only two stacks. The implemented queue should support all the functions of a normal queue (`push`, `peek`, `pop`, and `empty`).

```python
class MyQueue:

    def __init__(self):
        self.in_stack = []
        self.out_stack = []

    def push(self, x: int) -> None:
        self.in_stack.append(x)

    def _transfer(self):
        # Transfer elements from in_stack to out_stack if out_stack is empty
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())

    def pop(self) -> int:
        self._transfer()
        if self.out_stack:
            return self.out_stack.pop()
        else:
            return None # Or raise an error for empty queue

    def peek(self) -> int:
        self._transfer()
        if self.out_stack:
            return self.out_stack[-1]
        else:
            return None # Or raise an error for empty queue

    def empty(self) -> bool:
        return not self.in_stack and not self.out_stack

# Test cases
myQueue = MyQueue()
myQueue.push(1) # queue is: [1]
myQueue.push(2) # queue is: [1, 2] (leftmost is front of the queue)
print(f"Peek: {myQueue.peek()}")  # return 1
print(f"Pop: {myQueue.pop()}")    # return 1, queue is [2]
print(f"Empty: {myQueue.empty()}") # return False
myQueue.push(3)
print(f"Pop: {myQueue.pop()}")    # return 2, queue is [3]
print(f"Pop: {myQueue.pop()}")    # return 3, queue is []
print(f"Empty: {myQueue.empty()}") # return True
```
